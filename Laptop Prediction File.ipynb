### What techniques can be utilized to detect and eliminate any outliers present in the dataset?

```python
## To Detect The Outlier We Can Check using Boxplot or Histogram Like...
```

```python
data.hist();
```

```python
# So Here Much Number Of Outliers In price Column of The Data...
```

```python
# To Reduce/Remove Such Outliers We Mainly Have Three Method Like Following...

# (1.) Z-Score Method...
# (2.) Boundry Value Analysis...
# (3.) Interquartile Range (IQR)
```

```python
# So Here We Try First Method... (Z-Score)...
```

```python
# Formula = x - Mean / Std...
```

```python
data["Price"].mean()
```

```python
data["Price"].std()
```

```python
# Now We Just apply The Formual Like...
```

```python
df1 = data[["Price"]]
```

```python
df1
```

```python
# Store The Value In "df2"

df2 = (df1 - df1.mean()) / df1.std() 
```

```python
df2
```

```python
# Here The Value Which Is More Than +3 and Less Than -3 Are Outliers....
```

```python
(df2 > 3) | (df2 < -3)
```

```python
# See The Data..

df2[(df2 > 3) | (df2 < -3)]
```

```python
# Return NaN So We Can Drop It Like...
```

```python
df2[(df2 > 3) | (df2 < -3)].dropna()
```

```python
# Here We Can see Outliers Data....
```

```python
# Also The Data hithout Along From Outliers Are...
```

```python
df2[~((df2 > 3) | (df2 < -3))].dropna()
```

```python
(df2[(df2 > 3) | (df2 < -3)].dropna()).count()
```

```python
# So From 1020 We Have 28 Outliers.... So There Are 992 Data Is Free From It...
```

### What is the recommended approach for dividing the dataset into training and testing sets?

```python
# Using TrainTest Splits We Can divide The Data In Train and Test Sets...
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, train_size = 0.8, random_state = 42)
```

```python
# Here We Do Basic Partition But Change Is Necessary After The Model Creation and Testing...
```

### Which machine learning model is most suitable for training the data, considering its characteristics?

```python
# Here By Default We Can Use Linearregression But To Check..
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
# Training Score...

model.score(x_train, y_train)
```

```python
# Testing Score...

model.score(x_test, y_test)
```

```python
# This Is Still Not Efficient..
```

```python
# So We Employ KFold Technique Like...
```

```python
from sklearn.model_selection import KFold
```

```python
k = KFold(n_splits = 5)
```

```python
# Suppose We Do The Split of 5....
```

```python
# Also We Need a Cross_value Checker... Like....
```

```python
from sklearn.model_selection import cross_val_score
```

```python
k
```

```python
# Now The Fold Is 5 and We Can See The Basic Score Using cross_val_score Like...
```

```python
cross_val_score(model, x, y, cv = k)
```

```python
# Here It Predict The Score of "LinearRegression Model" 5 Times....
```

```python
# So We Need To Average Them....

(cross_val_score(model, x, y, cv = k)).mean()
```

```python
# So Basically The Average Score Of Our Model Is 74....
```

```python
# Now What About Other Models.... May It Predict More Than Linearregression Also....
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
```

```python
lis = [DecisionTreeRegressor, RandomForestRegressor, SVR, KNeighborsRegressor]
```

```python
for j in range(2, 11):
    print(f"\nResults for n_splits={j}:\n")
    for i in lis:
        k = KFold(n_splits = j)
        print(i)
        print(cross_val_score(i(), x, y, cv = k).mean())
```

```python
# So Here Using kFold and cross_val_score We Do 10 Partition And Check The Score of Multiple Models..


# Where The RandomForest Do The Better Scoring Then Others So We Choose It..
```

```python
model1 = RandomForestRegressor()
```

```python
model1.fit(x_train, y_train)
```

```python
model1.score(x_train, y_train)
```

```python
model1.score(x_test, y_test)
```

```python
# But There Is More Variation Between Training and Testing Score...
```
